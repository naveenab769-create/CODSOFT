```python
# Install required libs
!pip install nltk seaborn wordcloud

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import string
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

from wordcloud import WordCloud, STOPWORDS
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from sklearn import metrics
from sklearn import svm
from sklearn.metrics import roc_curve, roc_auc_score
from sklearn.metrics import ConfusionMatrixDisplay

nltk.download('stopwords')
nltk.download('punkt')

# ---- LOAD DATA ----
# The original file 'emails.csv' was not found.
# Assuming you are working with the SMS Spam Collection Dataset (common for text classification),
# I've updated the code to load 'SMSSpamCollection.txt'.
# Please ensure 'SMSSpamCollection.txt' is uploaded to your Colab environment
# or replace "SMSSpamCollection.txt" with your correct filename.
# If your file is 'emails.csv' and you've uploaded it, please change "SMSSpamCollection.txt" back to "emails.csv".

data_frame = pd.read_csv("SMSSpamCollection.txt", sep='\t', names=['label', 'text'])
# Convert 'ham' (0) and 'spam' (1) labels to numerical values for consistency with the rest of the code
data_frame['spam'] = data_frame['label'].map({'ham': 0, 'spam': 1})
# Drop the original 'label' column as it's no longer needed
data_frame = data_frame.drop(columns=['label'])

# ---- BASIC INFO ----
print(data_frame.head())
print(data_frame.info())

# ---- FEATURE: LENGTH ----
data_frame['length'] = data_frame['text'].apply(len)

# ---- HISTOGRAM (MATCH IMAGE) ----
plt.figure(figsize=(8,5))
data_frame[data_frame['spam']==0]['length'].hist(bins=100, label='Ham')
data_frame[data_frame['spam']==1]['length'].hist(bins=100, label='Spam')
plt.legend()
plt.title('Distribution of Length of Email Text')
plt.xlabel('Length')
plt.show()

# ---- WORD COUNT DISTRIBUTION ----
ham_words = [len(word_tokenize(text)) for text in data_frame[data_frame['spam']==0]['text']]
spam_words = [len(word_tokenize(text)) for text in data_frame[data_frame['spam']==1]['text']]

plt.figure(figsize=(8,5))
sns.histplot(ham_words, kde=True, label='Ham')
sns.histplot(spam_words, kde=True, label='Spam')
plt.legend()
plt.title("Distribution of Number of Words")
plt.show()

# ---- MEAN WORD LENGTH ----
def mean_word_length(x):
    words = word_tokenize(x)
    return np.mean([len(w) for w in words]) if words else 0

ham_mean = data_frame[data_frame['spam']==0]['text'].apply(mean_word_length)
spam_mean = data_frame[data_frame['spam']==1]['text'].apply(mean_word_length)

plt.figure(figsize=(8,5))
sns.histplot(ham_mean, kde=True, label='Ham')
sns.histplot(spam_mean, kde=True, label='Spam')
plt.legend()
plt.title("Distribution of Mean Word Length")
plt.show()

# ---- TEXT CLEANING ----
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = ''.join([c for c in text if c not in string.punctuation])
    words = [w for w in text.split() if w.lower() not in stop_words]
    return ' '.join(words)

data_frame['clean_text'] = data_frame['text'].apply(clean_text)

# ---- VECTORIZATION ----
vectorizer = CountVectorizer(min_df=2)
X = vectorizer.fit_transform(data_frame['clean_text'])
y = data_frame['spam']

# ---- TRAIN TEST SPLIT ----
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# ---- SVM MODEL ----
model = svm.SVC(kernel='linear', C=2, probability=True)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

# ---- OUTPUT (MATCH IMAGE STYLE) ----
print("\nOutput:\n")
print(classification_report(y_test, y_pred))

print("Test set")
print("Accuracy Score:", metrics.accuracy_score(y_test, y_pred))
print("F1 Score:", metrics.f1_score(y_test, y_pred))
print("Recall:", metrics.recall_score(y_test, y_pred))
print("Precision:", metrics.precision_score(y_test, y_pred))

# ---- NORMALIZED CONFUSION MATRIX ----
cm = confusion_matrix(y_test, y_pred)
cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

print("\nNormalized confusion matrix")
print(cm_norm)

plt.figure(figsize=(5,4))
sns.heatmap(cm_norm, annot=True, cmap="Blues", fmt=".3f")
plt.title("Normalized confusion matrix")
plt.xlabel("Predicted label")
plt.ylabel("True label")
plt.show()

# ---- ROC CURVE ----
probs = model.predict_proba(X_test)[:,1]
fpr, tpr, _ = roc_curve(y_test, probs)
auc_score = roc_auc_score(y_test, probs)

plt.figure()
plt.plot(fpr, tpr, label="SVM (AUC = %.3f)" % auc_score)
plt.plot([0,1],[0,1],'--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.title("ROC Curve")
plt.show()
```
